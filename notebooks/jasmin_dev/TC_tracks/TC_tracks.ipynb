# Demo opening all TC tracking .csv files from the JASMIN S3 object store

```python
import pandas as pd
import s3fs
```

```python
jasmin_s3 = s3fs.S3FileSystem(
    anon=True,
    client_kwargs={'endpoint_url': 'https://hackathon-o.s3-ext.jc.rl.ac.uk'},
)
jasmin_s3.ls('sim-data/analysis/TC_tracks/')
```

```python
dfs = {}
for s3path in jasmin_s3.ls('sim-data/analysis/TC_tracks/'):
    fname, ext = s3path.split('/')[-1].split('.')
    if ext == 'csv':
        with jasmin_s3.open(s3path, 'rb') as f:
            dfs[fname] = pd.read_csv(f)
```
