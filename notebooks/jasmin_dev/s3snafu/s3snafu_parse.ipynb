```python
from pathlib import Path

import numpy as np
import pandas as pd
```

```python
def parse_line(l):
    s3url = l.split(' ')[-1]
    var, chunk = s3url.split('/')[-2:]
    return var, chunk

def filter_files(chunk):
    if chunk in {'.zattrs', '.zgroup', '.zmetadata', '.zarray'}:
        return False
    return True

def get_df2d(path):
    t = path.read_text()
    ls = [l for l in t.split('\n') if l]
    
    vars_chunks = [([var] + chunk.split('.')) for var, chunk in [parse_line(l) for l in ls] if filter_files(chunk)]
    
    df = pd.DataFrame(((vc[0], int(vc[1]), int(vc[2])) for vc in vars_chunks if len(vc) == 3), columns=['var', 'c1', 'c2'])
    return df

def get_df3d(path):
    t = path.read_text()
    ls = [l for l in t.split('\n') if l]
    
    vars_chunks = [([var] + chunk.split('.')) for var, chunk in [parse_line(l) for l in ls] if filter_files(chunk)]
    
    df = pd.DataFrame(((vc[0], int(vc[1]), int(vc[2]), int(vc[3])) for vc in vars_chunks if len(vc) == 4), columns=['var', 'c1', 'c2', 'c3'])
    return df
   
```

```python
unique_vars = df2d['var'].unique()

for var in unique_vars:
    c1arr = df2d[df2d['var'] == var].sort_values(by='c1').groupby('c1').count().index.values
    a = np.arange(c1arr[-1] + 1)
    missing = np.ones(c1arr[-1] + 1).astype(bool)
    missing[c1arr] = False
    if len(a[missing]):
        if len(a[missing]) >= 3:
            print(var, a[missing][[0, 1, -1]])
        else:
            print(var, a[missing][0:1])
    else:
        print(var, 'complete')
```
